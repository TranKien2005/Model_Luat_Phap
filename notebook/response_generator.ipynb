```python
# ================================================================
# ⬇️⬇️⬇️ THAY ĐỔI Ở ĐÂY ⬇️⬇️⬇️
# ================================================================

# Chọn models: nhập list số hoặc "all"
# Ví dụ: [0] hoặc [0, 2] hoặc "all"
MODEL_INDICES = [0]

# Chọn tasks: nhập list số hoặc "all"
# Ví dụ: [0, 1, 2] hoặc "all"
TASK_INDICES = "all"

# ================================================================
# ⬆️⬆️⬆️ THAY ĐỔI Ở ĐÂY ⬆️⬆️⬆️
# ================================================================
```

```python
# Setup - Cài đặt thư viện và clone repo
!pip install transformers torch accelerate bitsandbytes -q
!git clone https://github.com/TranKien2005/Model_Luat_Phap.git

import sys
sys.path.append('./Model_Luat_Phap/evaluator')
```

```python
# Định nghĩa danh sách
ALL_MODELS = [
    "Qwen/Qwen2.5-7B-Instruct",
    "google/gemma-2-9b-it",
    "mistralai/Mistral-7B-Instruct-v0.3",
    "microsoft/Phi-3-mini-4k-instruct",
    "tiiuae/falcon-7b-instruct"
]

ALL_TASKS = [
    "general_issue_binary",
    "case_type_classification",
    "judgment_outcome_prediction",
    "numerical_constraint_check",
    "eligibility_logic_verification",
    "clause_type_identification",
    "legal_entity_extraction",
    "functional_sentence_labelling",
    "argument_consistency_check",
    "reasoning_method_detection",
    "vn_legal_mcq"
]

# Xác định models và tasks từ lựa chọn
if MODEL_INDICES == "all":
    models = ALL_MODELS
else:
    models = [ALL_MODELS[i] for i in MODEL_INDICES if 0 <= i < len(ALL_MODELS)]

if TASK_INDICES == "all":
    tasks = ALL_TASKS
else:
    tasks = [ALL_TASKS[i] for i in TASK_INDICES if 0 <= i < len(ALL_TASKS)]

print(f"📦 MODELS ({len(models)}): {models}")
print(f"📋 TASKS ({len(tasks)}): {tasks}")
```

```python
# Chạy generate responses
from pathlib import Path
from responseGenerator import generate_responses
from llmClient import unload_model

for idx, model in enumerate(models, 1):
    print(f"\n{'='*60}")
    print(f"📦 MODEL {idx}/{len(models)}: {model}")
    print("="*60)
    
    generate_responses(
        data_dir=Path("./Model_Luat_Phap/data"),
        model_name=model,
        output_dir=Path("/kaggle/working/responses"),
        tasks=tasks
    )
    
    unload_model()
    print(f"✅ Done: {model}")

print("\n✅ HOÀN TẤT!")
```
